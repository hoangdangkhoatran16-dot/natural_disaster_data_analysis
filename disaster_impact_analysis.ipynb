```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor

# Tạo dữ liệu phân tích mẫu
np.random.seed(42)
n = 1000
df = pd.DataFrame({
    'Population_Density': np.random.uniform(50, 5000, n),
    'Infrastructure_Score': np.random.uniform(1, 10, n),
    'GDP_Per_Capita': np.random.uniform(1000, 50000, n),
    'Disaster_Duration_Days': np.random.randint(1, 30, n),
    'Warning_Lead_Time_Hours': np.random.uniform(1, 72, n)
})

# Biến mục tiêu: Mức độ tác động nhân đạo
y = (df['Population_Density'] * 1.8 -
     df['Infrastructure_Score'] * 250 -
     df['Warning_Lead_Time_Hours'] * 50 +
     np.random.normal(0, 400, n))
X = df

# Huấn luyện Random Forest
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)

# Trực quan hóa Feature Importance
importance_df = pd.DataFrame({
    'Feature': ['Population Density', 'Infrastructure Score', 'GDP Per Capita', 'Disaster Duration', 'Warning Lead Time'],
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)

plt.figure(figsize=(10, 5), dpi=300)
sns.barplot(x='Importance', y='Feature', data=importance_df, palette='magma')
plt.title('Feature Importance: Key Drivers of Disaster Human Impact', fontsize=12, fontweight='bold')
plt.xlabel('Importance Score')
plt.ylabel('Drivers')
plt.tight_layout()
plt.savefig('feature_importance.png')
plt.show()
```
